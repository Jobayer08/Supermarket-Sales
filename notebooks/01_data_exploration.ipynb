```python
print("=== মোট বিক্রির পরিসংখ্যান ===")
print(df['Sales'].describe())
```

```python
branch_sales = df.groupby('Branch')['Sales'].sum().sort_values(ascending=False)
print("\n=== শাখা অনুযায়ী মোট বিক্রি (উচ্চ থেকে নিম্ন) ===")
print(branch_sales)
```

```python
product_sales = df.groupby('Product line')['Sales'].sum().sort_values(ascending=False)
print("\n=== পণ্যライン অনুযায়ী মোট বিক্রি ===")
print(product_sales)
```

```python
gender_spending = df.groupby('Gender')['Sales'].mean()
print("\n=== লিঙ্গ অনুযায়ী গড় খরচ ===")
print(gender_spending)
```

```python
payment_count = df['Payment'].value_counts()
print("\n=== পেমেন্ট মেথড ব্যবহারের সংখ্যা ===")
print(payment_count)
```

```python
df['Day_of_Week'] = df['Date'].dt.day_name()
day_sales = df.groupby('Day_of_Week')['Sales'].sum().sort_values(ascending=False)
print("\n=== সপ্তাহের দিন অনুযায়ী মোট বিক্রি ===")
print(day_sales)
```

```python
print(df.head())
```

```python
pivot = df.pivot_table(values='Sales', index='Branch', columns='Product line', aggfunc='mean')
print("\n=== পিভট টেবিল: শাখা ও পণ্যライン অনুযায়ী গড় বিক্রি ===")
print(pivot)
```

```python
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
print("\n=== সংখ্যাসূচক কলামগুলোর মধ্যে সম্পর্ক (Correlation Matrix) ===")
print(df[numeric_cols].corr())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style("darkgrid")
plt.rcParams['figure.figsize'] = (10, 6)  # ডিফল্ট সাইজ বড় করা
plt.rcParams['font.size'] = 12
```

```python
plt.figure()
branch_sales = df.groupby('Branch')['Sales'].sum().sort_values(ascending=False)
sns.barplot(x=branch_sales.index, y=branch_sales.values, palette='viridis')
plt.title('Branch-wise Total Sales', fontsize=16)
plt.xlabel('Branch')
plt.ylabel('Total Sales ($)')
plt.tight_layout()
plt.savefig('../visualizations/01_branch_sales.png', dpi=300)
plt.show()
```

```python
plt.figure()
product_sales = df.groupby('Product line')['Sales'].sum().sort_values(ascending=False)
sns.barplot(x=product_sales.values, y=product_sales.index, palette='magma')  # Horizontal bar
plt.title('Product Line-wise Total Sales', fontsize=16)
plt.xlabel('Total Sales ($)')
plt.ylabel('Product Line')
plt.tight_layout()
plt.savefig('../visualizations/02_product_sales.png', dpi=300)
plt.show()
```

```python
plt.figure()
sns.histplot(df['Sales'], bins=20, kde=True, color='blue')
plt.title('Distribution of Total Sales', fontsize=16)
plt.xlabel('Total Sales ($)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('../visualizations/03_sales_distribution.png', dpi=300)
plt.show()
```

```python
plt.figure()
sns.scatterplot(x='Unit price', y='Sales', data=df, hue='Branch', alpha=0.7)
plt.title('Unit Price vs Total Sales (by Branch)', fontsize=16)
plt.xlabel('Unit Price ($)')
plt.ylabel('Total Sales ($)')
plt.legend(title='Branch')
plt.tight_layout()
plt.savefig('../visualizations/04_price_vs_total.png', dpi=300)
plt.show()
```

```python
plt.figure()
sns.boxplot(x='Branch', y='Sales', data=df, palette='Set2')
plt.title('Sales Distribution by Branch (Box Plot)', fontsize=16)
plt.xlabel('Branch')
plt.ylabel('Total Sales ($)')
plt.tight_layout()
plt.savefig('../visualizations/05_branch_boxplot.png', dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', square=True, linewidths=0.5)
plt.title('Correlation Matrix of Numeric Features', fontsize=16)
plt.tight_layout()
plt.savefig('../visualizations/06_correlation_heatmap.png', dpi=300)
plt.show()
```

```python
print(df.dtypes)
```

```python
string_columns = ['Branch', 'City', 'Customer type', 'Gender', 'Product line', 'Payment']
for col in string_columns:
    df[col] = df[col].str.strip()
```

```python
print("=== মোট বিক্রির পরিসংখ্যান ===")
print(df['Sales'].describe())
```

```python
branch_sales = df.groupby('Branch')['Sales'].sum().sort_values(ascending=False)
print("\n=== শাখা অনুযায়ী মোট বিক্রি (উচ্চ থেকে নিম্ন) ===")
print(branch_sales)
```

```python
product_sales = df.groupby('Product line')['Sales'].sum().sort_values(ascending=False)
print("\n=== পণ্যライン অনুযায়ী মোট বিক্রি ===")
print(product_sales)
```

```python
gender_spending = df.groupby('Gender')['Sales'].mean()
print("\n=== লিঙ্গ অনুযায়ী গড় খরচ ===")
print(gender_spending)
```

```python
payment_count = df['Payment'].value_counts()
print("\n=== পেমেন্ট মেথড ব্যবহারের সংখ্যা ===")
print(payment_count)
```

```python
df['Day_of_Week'] = df['Date'].dt.day_name()
day_sales = df.groupby('Day_of_Week')['Sales'].sum().sort_values(ascending=False)
print("\n=== সপ্তাহের দিন অনুযায়ী মোট বিক্রি ===")
print(day_sales)
```

```python
print(df.head())
```

```python
pivot = df.pivot_table(values='Sales', index='Branch', columns='Product line', aggfunc='mean')
print("\n=== পিভট টেবিল: শাখা ও পণ্যライン অনুযায়ী গড় বিক্রি ===")
print(pivot)
```

```python
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
print("\n=== সংখ্যাসূচক কলামগুলোর মধ্যে সম্পর্ক (Correlation Matrix) ===")
print(df[numeric_cols].corr())
```
